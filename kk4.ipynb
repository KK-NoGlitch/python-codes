```python
def prime(a):
  sum=0
  for i in range (a):
    count=0
    for j in range (i+1):
      if (i+1)%(j+1)==0 :
        count+=1
    if count==2 :
      sum+=(i+1)
  print(sum)
num=int(input())
prime(num)
```
