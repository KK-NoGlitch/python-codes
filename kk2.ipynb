```python
money=10000
while True :
  print("1:ฝากเงิน 2:ถอนเงิน 3:ตรวจสอบเงิน 4:ออกจากระบบ")
  system=int(input())
  if system==1 :
    add=int(input("ฝากเท่าไหร่ :"))
    if add<200 :
      print("ต้องฝากอย่างน้อย 200")
    else :
      money+=add
  elif system==2 :
    sub=int(input("ถอนเท่าไหร่ :"))
    if sub<0 or sub>money :
      print("ERROR")
    else :
      money-=sub
  elif system==3 :
    print("คุณมีเงินในธนาคาร", money)
  elif system==4 :
    break
  else :
    print("ERROR")
print("คุณมีเงินในธนาคาร", money)
```

```python
NumG=0
Number=73
win=0
print("Guess a number between 1-100")
while NumG!=5 :
  guess=int(input())
  if guess<1 or guess>100 :
    print("Try again.")
  elif guess<Number and guess>=1 :
    print("Too low.")
    NumG+=1
  elif guess>Number and guess<=100 :
    print("Too high.")
    NumG+=1
  elif guess==Number :
    print("You're correct!")
    win=1
    break
if win==1 :
  print("You win!")
else :
  print("You lose")
```

```python
i=int(input("Enter a positive integer not more than 50 :"))
check=0
prime=0
if i>50 or i<1 :
  print("ERROR")
else :
  for j in range (i+1) :
    check=0
    for k in range (j) :
      if j%(k+2)==0 :
        check+=1
    if check==1 :
      prime+=1
  print(prime)
```
