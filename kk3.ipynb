```python
a=int(input())
b=int(input())
c=int(input())
d=int(input())
e=int(input())
f=int(input())
g=int(input())
h=int(input())
delcount=0
l=[a,b,c,d,e,f,g,h]
for i in range(8) :
  for j in range(8) :
    if i!=j and l[i]==l[j] :
      rem=l[i]
      l.remove(rem)
      l.append("T")
for k in range(8) :
  if l[k]=="T" :
    delcount+=1
for m in range(delcount) :
  l.remove("T")
print(l)
```

```python
amount=int(input())
cost=float(input())
member=input()
percent=0
print("รวม ",amount*cost)
if member=="y" :
  if amount*cost<=500 :
    percent=10
  elif amount*cost>500 and amount*cost<1000 :
    percent=15
  else :
    percent=20
if member=="n" :
  if amount*cost<=500 :
    percent=5
  elif amount*cost>500 and amount*cost<1000 :
    percent=10
  else :
    percent=15

print("ส่วนลด",amount*cost*percent/100)
print("จำนวนเงิน",amount*cost*(100-percent)/100)
```
